```python

# simple filter normalization
FILTER_MAP = {
    "ztfg":"g","ztfr":"r","ztfi":"i",
    "sdssu":"u","sdssg":"g","sdssr":"r","sdssi":"i","sdssz":"z",
    "ps1_g":"g","ps1_r":"r","ps1_i":"i","ps1_z":"z","ps1_y":"y",
    # Johnson-Cousins -> coarse mapping for the ML handoff
    "b":"u","v":"g","r":"r",
    "u":"u","g":"g","i":"i","z":"z","y":"y",
    "R":"r","V":"g","B":"u",
}

def _clean_header(cols):
    cleaned = []
    for c in cols:
        # normalize unicode; strip BOM & weird whitespace
        if not isinstance(c, str):
            c = str(c)
        c = unicodedata.normalize("NFKC", c)
        c = c.replace("\ufeff", "")  # BOM
        c = c.strip()
        # collapse internal whitespace runs to single space, then remove spaces
        c = re.sub(r"\s+", " ", c)
        # canonicalize a few expected labels
        c = c.replace("Mag Err", "MagErr")
        cleaned.append(c)
    return cleaned

def read_supernova_table_txt(path: Path) -> pd.DataFrame:
    """
    Tolerant reader for Sebastian's per-object .txt files.
    Uses a whitespace regex separator and cleans headers.
    """
    # Use explicit regex separator, python engine, ignore comments
    df = pd.read_csv(
        path, 
        sep=r"\s+", 
        engine="python", 
        comment="#", 
        dtype=str,   # read everything as string first; coerce later
        skip_blank_lines=True
    )
    df.columns = _clean_header(df.columns)
    return df

def coerce_bool(series: pd.Series) -> pd.Series:
    s = series.astype(str).str.strip().str.lower()
    return s.isin(["1","true","t","yes","y"])

def norm_filter(x):
    key = str(x).strip()
    low = key.lower()
    return FILTER_MAP.get(low, key)

def to_export(df_raw: pd.DataFrame) -> pd.DataFrame:
    """
    Convert a per-object table to Felipe's 5 columns:
      mjd, mag, mag_err, filter, detected
    Interprets UL=True as non-detection -> detected=0, mag_err=inf.
    Treats MagErr = -1.0 as 'unknown' (kept as NaN) for detections; for UL rows we set inf anyway.
    """
    # columns we expect in these tables
    need = ["MJD","Mag","MagErr","Filter","UL"]
    cols = {c: c for c in df_raw.columns}
    missing = [c for c in need if c not in cols]
    if missing:
        # show for debugging
        raise ValueError(f"Missing columns {missing}. Found: {list(df_raw.columns)}")

    df = df_raw.copy()

    # coerce numerics
    mjd = pd.to_numeric(df["MJD"], errors="coerce")
    mag = pd.to_numeric(df["Mag"], errors="coerce")

    # MagErr handling:
    # - Many ROTSE UL rows use -1.0; we will coerce to NaN first
    mag_err = pd.to_numeric(df["MagErr"], errors="coerce").astype(float)
    mag_err[mag_err < 0] = np.nan

    # UL flag: True means upper limit
    ul = coerce_bool(df["UL"])

    # detected flag
    detected = (~ul).astype(np.int8)

    # for UL rows, enforce Felipe convention
    mag_err[ul.values] = np.inf

    # filters normalized
    filt = df["Filter"].map(norm_filter).astype("string")

    out_perevent = pd.DataFrame({
        "mjd": mjd.astype(float),
        "mag": mag.astype(float),
        "mag_err": mag_err,
        "filter": filt,
        "UL": ul,
        "detected": detected
    }).sort_values("mjd").reset_index(drop=True)

    # drop rows that are totally empty (all NaN except 'detected' which would be 1 by default if UL missing)
    keep = out_perevent[["mjd","mag","mag_err","UL","filter"]].notna().any(axis=1)
    out_perevent = out_perevent[keep].reset_index(drop=True)

    return out_perevent[["mjd","mag","mag_err","UL", "filter","detected"]]
```

```python
# Cell 3 — process a single event for quick sanity check
def process_one(event_name: str):
    event_dir = supernovae_dir / event_name
    infile = event_dir / f"{event_name}.txt"
    if not infile.exists():
        raise FileNotFoundError(f"Could not find {infile}")

    # use the reader you defined above
    df_raw = read_supernova_table_txt(infile)
    df_out = to_export(df_raw)

    # write parquet/csv to the *lowercase* 'out_perevent' folder you created
    if write_parquet:
        df_out.to_parquet(out_perevent / f"{event_name}.parquet",
                          index=False, engine="pyarrow",
                          compression="zstd", compression_level=7)
    if write_csv:
        df_out.to_csv(out_perevent / f"{event_name}.csv", index=False)

    return df_out
```

```python
def load_allparams_robust(path):
    # Grab the first non-comment line as header text
    with open(path, "r", encoding="utf-8", errors="replace") as f:
        for line in f:
            if line.strip() and not line.lstrip().startswith("#"):
                header_line = line.rstrip("\n")
                break
    # Split header on 2+ spaces OR tabs
    cols = re.split(r"\s{2,}|\t+", header_line.strip())
    # Now read the whole file using the same rule; skip the header line we just consumed
    df = pd.read_csv(
        path,
        sep=r"\s{2,}|\t+",
        engine="python",
        header=0,          # header is the first line in the file
        comment="#",
        dtype=str,         # start as strings; coerce later as needed
        skip_blank_lines=True,
        on_bad_lines="warn"  # or "skip" to silently drop malformed rows
    )
    # Ensure we got the expected header width; if not, set names explicitly
    if len(df.columns) != len(cols):
        df = pd.read_csv(
            path,
            sep=r"\s{2,}|\t+",
            engine="python",
            header=None,
            names=cols,
            comment="#",
            dtype=str,
            skip_blank_lines=True,
            on_bad_lines="warn"
        )
    return df

allparams_df = load_allparams_robust(allparams_path)
print("Loaded shape:", allparams_df.shape)
```

```python
def resolve_cols(df, requested=None):
    cols = list(df.columns)
    lower = {c.lower(): c for c in cols}
    anchor = lower.get("name", cols[0])  # prefer 'name'; else first
    out = [anchor]
    if requested is None:
        return out
    if not isinstance(requested, (list, tuple)):
        requested = [requested]
    for r in requested:
        if isinstance(r, int):
            if 0 <= r < len(cols) and cols[r] not in out:
                out.append(cols[r]); continue
        key = str(r).strip().lower()
        if key in lower:
            c = lower[key]
            if c not in out: out.append(c)
        else:
            # soft partial match
            for c in cols:
                if key in c.lower() and c not in out:
                    out.append(c)
    return out

def print_cols(df, requested=None, head=10):
    show = resolve_cols(df, requested)
    display(df[show].head(head))
    return show
```

```python
# Preview just the anchor column
print_cols(allparams_df, None, head=10)

selected_cols = [1, 2, 3]  # indices (0-based)

# Resolve to real column names (this prepends the anchor: 'name' or first column)
resolved = resolve_cols(allparams_df, selected_cols)
df_cols  = allparams_df[resolved].copy()

# Optional: coerce numeric columns (skip the anchor at index 0)
for c in resolved[1:]:
    df_cols[c] = pd.to_numeric(df_cols[c], errors="coerce")

# Build stub from the *names* associated with your indices (exclude anchor)
idx_names = [allparams_df.columns[i] for i in selected_cols if 0 <= i < len(allparams_df.columns)]

import re
def sanitize(s): 
    return re.sub(r"[^A-Za-z0-9._-]+", "_", s)

stub = "_".join(sanitize(c) for c in idx_names) if idx_names else "cols"

# Write parquet/csv
if write_parquet:
    df_cols.to_parquet(
        out / f"allevent_{stub}.parquet",
        index=False, engine="pyarrow",
        compression="zstd", compression_level=7
    )
if write_csv:
    df_cols.to_csv(out / f"allevent_{stub}.csv", index=False)

display(df_cols.head(10))
print("Columns used (resolved):", resolved)
print("Filename stub:", stub)
```
